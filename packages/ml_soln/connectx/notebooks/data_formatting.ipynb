```python
import tensorflow as tf
import random
```

```python
def generate_board(n_rows=6, n_cols=7):
    values = (0, 1, 2)
    return [
        [random.choice(values) for _ in range(n_cols)]
        for _ in range(n_rows)
    ]
```

```python
board = generate_board()
board = tf.convert_to_tensor(board)
board
```

```python
flat = tf.reshape(board, shape=(-1))
flat
```

```python
boards = [
    tf.cast(flat == 1, dtype='float16'),
    tf.cast(flat == 2, dtype='float16')
]
boards = tf.convert_to_tensor(boards)
boards
```

```python
from keras.api.layers import Input, Lambda, Reshape
from keras.api.models import Model

n_rows = 6
n_cols = 7

def split_player_boards(flat_board):
    return tf.stack([
        # player 1's moves
        tf.cast(flat_board == 1, dtype='float16'),
        # player 2's moves
        tf.cast(flat_board == 2, dtype='float16')
    ], axis=-1)

board_input = Input(shape=(n_rows * n_cols,),
                    name='board')
board_layers = Lambda(function=split_player_boards,
                      output_shape=(n_rows * n_cols, 2))(board_input)
board_layers = Reshape(target_shape=(n_rows, n_cols, 2))(board_layers)

model = Model(inputs=[board_input], outputs=[board_layers])
```
